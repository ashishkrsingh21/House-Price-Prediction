```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
house_data = pd.read_csv("/Users/ashishkrsingh/Developers/Machine Learning/Projects/House-Price-Prediction/data/train.csv")
```

```python
print(f"House data shape: {house_data.shape}")
print(house_data.info())
```

```python
house_data.head(10)
```

```python
print(house_data["City"].unique())
print(house_data["Locality_Type"].unique())
print(house_data["Furnishing_Status"].unique())
print(house_data["Property_Type"].unique())
```

```python
print(house_data["City"].value_counts())
print(house_data["Locality_Type"].value_counts())
print(house_data["Furnishing_Status"].value_counts())
print(house_data["Property_Type"].value_counts())
```

```python
print(house_data["Price_INR_Lakhs"].describe())
```

```python
plt.figure(figsize=(10, 6))
plt.hist(house_data["Price_INR_Lakhs"], bins=50, color='skyblue', edgecolor='black')
plt.title("Distribution of House Prices (in Lakhs INR)")
plt.xlabel("Price (INR Lakhs)")
plt.ylabel("Frequency")
plt.show()
```

```python
house_data.hist(figsize=(12, 10), bins=30, color='lightgreen', edgecolor='black')
plt.tight_layout()
plt.show()
```

```python
correlation = house_data.select_dtypes(include="number").corr()
print(correlation["Price_INR_Lakhs"].sort_values(ascending=False))
```

```python
plt.Figure(figsize=(10, 6))
plt.scatter(house_data["Super_Area_SqFt"],house_data["Price_INR_Lakhs"], alpha=0.3, color='orange')
plt.xlabel("Super Area (Sqft)")
plt.ylabel("Price (INR Lakhs)")
plt.title("Price vs Super Area")  
plt.show()
```

```python
plt.Figure(figsize=(10, 6))
house_data.boxplot(column="Price_INR_Lakhs", by="City", grid=True)
plt.xlabel("City")
plt.ylabel("Price (INR Lakhs)")
plt.title("Price Distribution by City") 
plt.suptitle("")  # Suppress the default title to avoid overlap
plt.xticks(rotation=45)
plt.show()
```

```python
feature_corr = house_data.select_dtypes(include="number").corr()

plt.figure(figsize=(12, 9))

plt.imshow(feature_corr, cmap="coolwarm", aspect="auto")

plt.xticks(
    range(len(feature_corr.columns)),
    feature_corr.columns,
    rotation=90
)

plt.yticks(
    range(len(feature_corr.columns)),
    feature_corr.columns
)

plt.colorbar(label="Correlation")

plt.title("Correlation Matrix of Numerical Features")

plt.tight_layout()
plt.show()
```
